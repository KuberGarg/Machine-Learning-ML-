```python
import pandas as pd
```

```python
df=pd.read_csv("loan data.csv")
```

```python
df.head()
```

# Unique Categories

```python
df["Property_Area"].unique()
```

# Category Order

```python
en_data_order=[[ 'Rural', 'Semiurban','Urban']]
```

# Apply Ordinal Encoding

```python
from sklearn.preprocessing import OrdinalEncoder
```

```python
oe= OrdinalEncoder(categories=en_data_order)
df["Property_Area"]=oe.fit_transform(df[["Property_Area"]])
```

```python
df.head()
```
