```python
import pandas as pd
import numpy as np
```

```python
dataset = pd.read_csv("supermarket.csv")
```

```python
dataset.head(3)
```

```python
market = []
for i in range(0, dataset.shape[0]):
    cus = []
    for j in dataset.columns:
        if type(dataset[j][i]) == str:
            cus.append(dataset[j][i])
    market.append(cus)
```

```python
market[:5]
```

```python
l = []
for i in market:
    for j in i:
        l.append(j)
```

```python
import collections
```

```python
p = collections.Counter(l)
```

```python
d = {
    "Item Name": p.keys(),
    "values": p.values()
}
```

```python
pd.DataFrame(d).sort_values(by=["values"], ascending=False)
```

```python
from mlxtend.preprocessing.transactionencoder import TransactionEncoder
```

```python
tr = TransactionEncoder()
tr.fit(market)
data = tr.transform(market)
```

```python
df = pd.DataFrame(data, columns=tr.columns_)
```

```python
df.head()
```

```python
from mlxtend.frequent_patterns import fpgrowth
```

```python
fpgrowth(df,min_support=0.10,use_colnames=True,max_len=4).sort_values(by=["support"])
```
