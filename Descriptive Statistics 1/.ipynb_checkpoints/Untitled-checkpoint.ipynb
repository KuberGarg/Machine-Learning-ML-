# Mean Absolute Deviation

```python
a= np.array([75,65,73,68,72,76])
b=np.array([90,47,43,96,93,51])
no=np.array([1,2,3,4,5,6])
```

```python
mean=np.mean(b)
mean
```

```python
mad_a=np.sum(np.abs(a-mean))/len(a)
mad_b=np.sum(np.abs(b-mean))/len(b)
```

```python
mad_a,mad_b
```
