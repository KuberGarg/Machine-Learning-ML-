```python
import scipy.stats as st
import numpy as np
```

```python
old_design_data =np.array([
    45.2, 42.8, 38.9, 43.5, 41.0, 44.6, 40.5, 42.7, 39.8, 41.4,
    44.3, 39.7, 42.1, 40.6, 43.0, 42.2, 41.5, 39.6, 44.0, 43.1,
    38.7, 43.9, 42.0, 41.9, 42.8, 43.7, 41.3, 40.9, 42.5, 41.6
])

new_design_data = np.array([
    48.5, 49.1, 50.2, 47.8, 48.7, 49.9, 48.0, 50.5, 49.8, 49.6,
    48.2, 48.9, 49.7, 50.3, 49.4, 50.1, 48.6, 48.3, 49.0, 50.0,
    48.4, 49.3, 49.5, 48.8, 50.6, 50.4, 48.1, 49.2, 50.7, 50.1
])
```

```python
pop_std=2.5
n_sample=len(new_design_data)
ap= 0.05
```

```python
mean_new =np.mean(new_design_data)
mean_old =np.mean(old_design_data)
```

```python
z_cal=(mean_new -mean_old)/(pop_std/np.sqrt(n_sample))
```

```python
z_table=st.norm.ppf(1-ap)
z_table
```

```python
if z_cal>z_table:
    print("Ha is right")
else:
    print("H0 is right")
    
```
