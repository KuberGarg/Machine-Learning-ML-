```python
import scipy.stats as st
import numpy as np
```

```python
Team_A = np.array([75, 80, 85, 78, 82, 79, 81, 83, 77, 80, 84, 76, 82, 85, 79, 81, 80, 78, 83, 82])

Team_B = np.array([70, 75, 78, 73, 76, 74, 77, 72, 75, 76,79, 71, 74, 77, 73, 75, 76, 72, 78, 74])
```

```python
std_a = np.std(Team_A)
std_b = np.std(Team_B)

mean_a = np.mean(Team_A)
mean_b = np.mean(Team_B)
```

```python
t_cal = (mean_a-mean_b) / (np.sqrt(((std_a*std_a)/len(Team_A))+((std_b*std_b)/len(Team_B))))
t_cal
```
